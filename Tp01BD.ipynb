```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

base_url = "https://www.gutenberg.org/ebooks/search/?sort_order=downloads&start_index="
books = []

for page in range(0, 1000, 25):  # start_index=0,25,50,...
    url = f"{base_url}{page}"
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")

    # كل كتاب ضمن <li class="booklink">
    for li in soup.select("li.booklink"):
        title = li.select_one("span.title")
        author = li.select_one("span.subtitle")
        link_tag = li.find("a")

        title = title.text.strip() if title else "N/A"
        author = author.text.strip() if author else "Unknown"
        link = "https://www.gutenberg.org" + link_tag["href"] if link_tag else ""

        books.append({
            "Title": title,
            "Author": author,
            "Link": link
        })

    print(f" Page {page//25 + 1} done, total books: {len(books)}")
    time.sleep(0.3)

df = pd.DataFrame(books)
df.to_csv("gutenberg_books.csv", index=False, encoding="utf-8")
print(f"\n تم حفظ {len(df)} كتاب في الملف 'gutenberg_books.csv'")
```
