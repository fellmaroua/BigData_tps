# Step 2: Read the log file
قراءة ملف السجلات

```python
with open("weblogs.txt", "r") as f:
    lines = f.readlines()

lines[:3]
```

# Step 3: Mapper function – extract HTTP status code


تعريف دالة الـ Mapper

```python
def mapper(line):
    parts = line.strip().split(",")

    status = parts[5]

    return (status, 1)
```

# Step 4: Apply Map function to all log lines



```python
mapped_data = []

for line in lines:
    mapped_data.append(mapper(line))

mapped_data
```

# Step 5: Group values by status code (Shuffle phase)


```python
from collections import defaultdict

grouped_data = defaultdict(list)

for status, count in mapped_data:
    grouped_data[status].append(count)

grouped_data
```

# Step 6: Reducer function – sum counts


```python
def reducer(status, values):
    return sum(values)
```

# Step 7: Apply Reduce function to grouped data


```python
final_result = {}

for status, values in grouped_data.items():
    final_result[status] = reducer(status, values)

final_result
```

# Step 8: Display number of requests per HTTP status code


```python
for status, count in sorted(final_result.items()):
    print(f"HTTP {status}: {count} requests")
```
