```python
import requests
from bs4 import BeautifulSoup
import csv

base_url = "https://books.toscrape.com/catalogue/page-{}.html"
headers = {"User-Agent": "Mozilla/5.0"}

all_books = []

page = 1
while True:
    url = base_url.format(page)
    resp = requests.get(url, headers=headers)

    if resp.status_code != 200:
        print(f" توقف عند الصفحة {page} (غير موجودة).")
        break

    soup = BeautifulSoup(resp.text, "html.parser")
    books = soup.select("article.product_pod")

    if not books:
        break

    print(f" استخراج الصفحة {page} ...")

    for b in books:
        title = b.h3.a["title"]
        price = b.select_one("p.price_color").get_text(strip=True)
        availability = b.select_one("p.availability").get_text(strip=True)

        all_books.append([title, price, availability])

    page += 1
with open("books_all.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Title", "Price", "Availability"])
    writer.writerows(all_books)

print(" تم استخراج جميع الكتب وحفظها في books_all.csv")
```
